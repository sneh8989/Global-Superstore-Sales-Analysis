## Business Context
Global Superstore is a global e-commerce retail dataset tracking customer orders from 2011 to 2014 across 147 countries, 3 product categories, and multiple customer segments. The goal of this phase is to establish ground truth data profiles prior to cleaning and transformation.


```python
import os
import pandas as pd
import numpy as np

# Resolve path portably
data_path = "../data/raw/superstore.csv" if os.path.exists("../data/raw/superstore.csv") else "data/raw/superstore.csv"
df_raw = pd.read_csv(data_path)
print(f"Dataset successfully loaded. Dimensions: {df_raw.shape[0]:,} rows x {df_raw.shape[1]} columns")
```

## 3. Missing Value & Duplicate Audit


```python
null_counts = df_raw.isnull().sum()
duplicate_rows = df_raw.duplicated().sum()
print("Missing values per column:")
print(null_counts[null_counts > 0] if (null_counts > 0).any() else "Zero missing values detected across all columns.")
print(f"Exact duplicate rows: {duplicate_rows}")
```

## 4. Categorical Cardinality & Unique Values
Inspect unique entity counts across primary business dimensions.


```python
cardinality = {
    "Categories": df_raw["Category"].nunique(),
    "Sub-Categories": df_raw["Sub.Category"].nunique(),
    "Markets": df_raw["Market"].nunique(),
    "Regions": df_raw["Region"].nunique(),
    "Countries": df_raw["Country"].nunique(),
    "Customer Segments": df_raw["Segment"].nunique(),
    "Unique Customers": df_raw["Customer.ID"].nunique(),
    "Unique Orders": df_raw["Order.ID"].nunique(),
    "Unique Products": df_raw["Product.ID"].nunique(),
}
pd.DataFrame(list(cardinality.items()), columns=["Dimension", "Unique Count"])
```

## 6. Data Quality Observations & Action Plan
1. **Zero Null Values:** The dataset is structurally complete with 51,290 records.
2. **Tableau Artifact Column:** The column `记录数` ('Number of Records') contains constant `1` for all rows and should be dropped.
3. **Date Formats:** `Order.Date` and `Ship.Date` are ISO timestamp strings and must be parsed into datetime objects.
4. **Column Naming:** Dot notation (`Customer.ID`, `Order.Date`, etc.) is maintained to preserve compatibility with the Power BI dashboard data model.
5. **Feature Opportunities:** Fulfillment duration (`Shipping_Days`), `Unit_Price`, `Profit_Margin_Pct`, and `Discount_Band` will be engineered in the cleaning pipeline.
