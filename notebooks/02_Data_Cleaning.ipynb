## Pipeline Overview
1. Load raw dataset (`data/raw/superstore.csv`).
2. Remove Tableau artifact column `记录数`.
3. Parse `Order.Date` and `Ship.Date` into datetime format.
4. Engineer temporal attributes (`Year`, `Quarter`, `Month`, `Month_Name`, `YearMonth`, `Day_of_Week`).
5. Engineer operational metrics (`Shipping_Days`, `Unit_Price`, `Profit_Margin_Pct`, `Estimated_Cost`, `Discount_Band`).
6. Export cleaned dataset to `data/processed/cleaned_superstore.csv` maintaining Power BI schema parity.


```python
import os
import sys
import pandas as pd
import numpy as np

# Ensure scripts module is accessible
sys.path.insert(0, os.path.abspath(".."))
from scripts.data_cleaning import load_raw_data, clean_superstore_data, save_cleaned_data

raw_path = "../data/raw/superstore.csv" if os.path.exists("../data/raw/superstore.csv") else "data/raw/superstore.csv"
output_path = "../data/processed/cleaned_superstore.csv" if os.path.exists("../data/raw/superstore.csv") else "data/processed/cleaned_superstore.csv"

raw_df = load_raw_data(raw_path)
cleaned_df, report = clean_superstore_data(raw_df)
save_cleaned_data(cleaned_df, output_path)

print("Data cleaning executed successfully!")
for k, v in report.items():
    print(f"  {k}: {v}")
```
