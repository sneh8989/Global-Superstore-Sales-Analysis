**Objective:** Cross-tabulate metrics across product categories, customer segments, regions, and markets, and generate publication-grade Excel workbooks.


```python
import os
import sys
import pandas as pd

sys.path.insert(0, os.path.abspath(".."))
from scripts.pivot_exporter import export_all_pivot_tables_and_reports

data_path = "../data/processed/cleaned_superstore.csv" if os.path.exists("../data/processed/cleaned_superstore.csv") else "data/processed/cleaned_superstore.csv"
df = pd.read_csv(data_path)
```

## 1. Sales by Category & Customer Segment


```python
pd.pivot_table(
    df, index="Category", columns="Segment", values="Sales", aggfunc="sum", margins=True, margins_name="Total Revenue"
).round(2)
```

## 2. Profit by Category & Customer Segment


```python
pd.pivot_table(
    df, index="Category", columns="Segment", values="Profit", aggfunc="sum", margins=True, margins_name="Total Profit"
).round(2)
```

## 3. Profit by Region & Product Category


```python
pd.pivot_table(
    df, index="Region", columns="Category", values="Profit", aggfunc="sum", margins=True, margins_name="Total Profit"
).round(2)
```

## 4. Profit by Market & Product Category


```python
pd.pivot_table(
    df, index="Market", columns="Category", values="Profit", aggfunc="sum", margins=True, margins_name="Total Profit"
).round(2)
```
