## 2. Quantifying the Discount Bleed
Analyze transactions with discounts > 20% vs. <= 20%.


```python
high_disc = df[df["Discount"] > 0.20]
safe_disc = df[df["Discount"] <= 0.20]

print(f"High Discount (>20%) Line Items : {len(high_disc):,} items across {high_disc['Order.ID'].nunique():,} unique orders")
print(f"High Discount Total Sales       : ${high_disc['Sales'].sum():,.2f}")
print(f"High Discount Total Profit      : ${high_disc['Profit'].sum():,.2f}")
print(f"Safe Discount (<=20%) Line Items: {len(safe_disc):,} items across {safe_disc['Order.ID'].nunique():,} unique orders")
print(f"Safe Discount (<=20%) Total Sales : ${safe_disc['Sales'].sum():,.2f}")
print(f"Safe Discount (<=20%) Total Profit: ${safe_disc['Profit'].sum():,.2f}")
print(f"Safe Discount Profit Margin     : {(safe_disc['Profit'].sum()/safe_disc['Sales'].sum()*100):.2f}%")
```

## 3. Executive Strategic Recommendations

### Pillar 1: Enforce Strict Discount Governance
- **Recommendation:** Implement automated discount thresholds capping standard salesperson discounts at 15%. Require VP approval for discounts above 20%.
- **Impact:** Eliminates up to $500,000+ in unnecessary margin destruction across 12,500+ loss-making line items.

### Pillar 2: Overhaul the Tables Sub-Category
- **Recommendation:** Increase minimum order quantities on Tables, renegotiate freight supplier terms, or eliminate low-margin SKUs (e.g. Cubify, Chromcraft tables).
- **Impact:** Converts a -$64,083 drag into break-even or positive operational cash flow.

### Pillar 3: Scale High-Margin Technology Offerings
- **Recommendation:** Direct marketing and enterprise sales campaigns toward high-performing Technology lines (Copiers, Phones, Accessories).
- **Impact:** Expands high-margin revenue (14%+) without increasing working capital intensity.

### Pillar 4: Restructure Underperforming Regions
- **Recommendation:** Review distributor contracts and freight cost allocations in Southeast Asia and EMEA where Furniture profit margins are negative.
- **Impact:** Eliminates regional profit subsidies and optimizes global supply chain efficiency.

### Pillar 5: Customer Retention & VIP Programs
- **Recommendation:** Establish a dedicated account management program for the top 100 enterprise customers who contribute disproportionate revenue and recurring volume.
- **Impact:** Protects high-value recurring revenue streams while improving net promoter score (NPS).
